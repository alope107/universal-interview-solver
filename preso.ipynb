````python
def odd_or_even(num):
    if num % 2 == 1:
        return "odd"
    else:
        return "even"

assert odd_or_even(7) == "odd"
assert odd_or_even(22) == "even"
assert odd_or_even(-3) == "odd"
print("All tests passed!")
````

````python
def reverse_string(s):
    return "".join(reversed(s))

assert reverse_string("hello") == "olleh"
assert reverse_string("Sweden") == "nedewS"
assert reverse_string("") == ""
print("All tests passed!")
````
